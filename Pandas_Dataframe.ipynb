```python
#Q1
#Create a Pandas DataFrame using a Python dictionary with the following columns:

import pandas as pd

data = {
    "product": ["laptop", "mobile", "speaker", "keyboard", "power bank"],
    "region": ["north", "west", "west", "south", "east"],
    "sales": [200000, 150000, 55000, 9000, 14000],
    "quantity": [6, 8, 5, 13, 18]
}

df = pd.DataFrame(data)
print(df)
```
